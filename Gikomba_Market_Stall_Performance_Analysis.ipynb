## 1. Introduction

This project analyses 14 days of daily sales and stock records for Wanjiku's fresh-produce stall at Gikomba Market.

The objective is to evaluate business performance, identify sales trends, classify daily performance, assess stock levels and calculate product gross margins.

The analysis uses core Python concepts, including lists, dictionaries, functions, loops, conditional statements and formatted output.

```python
# Daily sales for 14 days (KES)
daily_sales = [
    2300, 1950, 2800, 2100, 3600, 5200, 4100,
    2500, 2200, 3100, 2400, 3900, 5600, 4700
]

# Days of the week
days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"] * 2

# Stock information
stock = {
    "Tomatoes": {"qty": 18, "reorder": 25, "cost": 60, "price": 90},
    "Onions": {"qty": 40, "reorder": 20, "cost": 70, "price": 100},
    "Sukuma": {"qty": 9, "reorder": 15, "cost": 30, "price": 50},
    "Potatoes": {"qty": 55, "reorder": 30, "cost": 50, "price": 75},
    "Avocado": {"qty": 12, "reorder": 10, "cost": 80, "price": 140}
}

print("Dataset loaded successfully!")
```

## 2. Daily Sales Performance

This section calculates the total sales, average daily sales,
highest sales and lowest sales recorded over the 14-day period.

```python
# Calculate total sales
total_sales = sum(daily_sales)

# Calculate average daily sales
average_sales = total_sales / len(daily_sales)

print("Total Sales:", total_sales)
print("Average Daily Sales:", average_sales)
```

```python
# Find highest and lowest sales
highest_sales = max(daily_sales)
lowest_sales = min(daily_sales)

print("Highest Sales:", highest_sales)
print("Lowest Sales:", lowest_sales)
```

```python
# Find the positions of the highest and lowest sales
best_index = daily_sales.index(highest_sales)
worst_index = daily_sales.index(lowest_sales)

# Identify the corresponding days
best_day = days[best_index]
worst_day = days[worst_index]

# Identify the week numbers
best_week = (best_index // 7) + 1
worst_week = (worst_index // 7) + 1

print("Best Performing Day:", best_day)
print("Best Sales Amount:", highest_sales)
print("Best Week:", best_week)

print()

print("Worst Performing Day:", worst_day)
print("Worst Sales Amount:", lowest_sales)
print("Worst Week:", worst_week)
```

```python
print("=" * 45)
print("       DAILY SALES PERFORMANCE")
print("=" * 45)

print(f"Total Sales:       KES {total_sales:,.2f}")
print(f"Average Sales:     KES {average_sales:,.2f}")
print(f"Highest Sales:     KES {highest_sales:,.2f}")
print(f"Lowest Sales:      KES {lowest_sales:,.2f}")

print("-" * 45)

print(f"Best Day:          {best_day} (Week {best_week})")
print(f"Worst Day:         {worst_day} (Week {worst_week})")

print("=" * 45)
```

## 3. Weekly Sales Analysis

This section calculates the total sales for each week and
determines the percentage change in sales between the two weeks.

```python
def week_totals(sales):
    """
    Calculate the total sales for Week 1 and Week 2.

    Args:
        sales: A list containing 14 daily sales figures.

    Returns:
        A tuple containing Week 1 and Week 2 sales totals.
    """

    week1 = sum(sales[:7])
    week2 = sum(sales[7:])

    return week1, week2
```

```python
# Call the function and unpack the returned values
week1_total, week2_total = week_totals(daily_sales)

print("Week 1 Sales:", week1_total)
print("Week 2 Sales:", week2_total)
```

## 4. Daily Sales Classification

This section classifies each day's sales as Strong, Normal or Weak
based on its relationship to the average daily sales.

```python
def classify_day(amount, average):
    """
    Classify daily sales based on the average sales.

    Args:
        amount: Sales amount for a particular day.
        average: Average daily sales.

    Returns:
        Strong, Normal or Weak.
    """

    if amount > average * 1.20:
        return "Strong"

    elif amount < average * 0.80:
        return "Weak"

    else:
        return "Normal"
```

```python
# Classify each day's sales
daily_classifications = []

for amount in daily_sales:
    category = classify_day(amount, average_sales)
    daily_classifications.append(category)

print(daily_classifications)
```

```python
# Initialize category counts
sales_count = {
    "Strong": 0,
    "Normal": 0,
    "Weak": 0
}

# Count the classifications
for category in daily_classifications:
    sales_count[category] += 1

# Display the results
print("DAILY SALES CLASSIFICATION")
print("-" * 35)

for category, count in sales_count.items():
    print(f"{category:<10}: {count} days")
```

```python
print("DAILY SALES BREAKDOWN")
print("-" * 50)

for i in range(len(daily_sales)):
    print(
        f"{days[i]:<8} "
        f"KES {daily_sales[i]:>8,} "
        f"{daily_classifications[i]}"
    )
```

## 5. Weekday vs Weekend Sales Analysis

This section compares average sales during weekdays and weekends
to identify differences in customer purchasing patterns.

```python
# Separate weekday sales
weekday_sales = daily_sales[:5] + daily_sales[7:12]

# Separate weekend sales
weekend_sales = daily_sales[5:7] + daily_sales[12:14]

# Calculate average sales
weekday_average = sum(weekday_sales) / len(weekday_sales)

weekend_average = sum(weekend_sales) / len(weekend_sales)

# Display results
print("Weekday Average Sales:", weekday_average)
print("Weekend Average Sales:", weekend_average)
```

## 6. Stock Management and Gross Margin Analysis

This section identifies products that require restocking and
calculates the gross margin percentage for each product.

```python
def products_to_restock(stock):
    """
    Identify products whose current quantity is at or below
    their reorder level.

    Args:
        stock: Dictionary containing product stock information.

    Returns:
        A list of product names requiring restocking.
    """

    restock_list = []

    for product, details in stock.items():

        if details["qty"] <= details["reorder"]:
            restock_list.append(product)

    return restock_list
```

```python
# Identify products requiring restocking
restock_products = products_to_restock(stock)

print("PRODUCTS REQUIRING RESTOCKING")
print("-" * 35)

for product in restock_products:
    print(product)
```

```python
def calculate_gross_margin(stock):
    """
    Calculate the gross margin percentage for every product.

    Args:
        stock: Dictionary containing product cost and selling price.

    Returns:
        Dictionary containing each product's gross margin percentage.
    """

    margins = {}

    for product, details in stock.items():

        cost = details["cost"]
        price = details["price"]

        margin = ((price - cost) / price) * 100

        margins[product] = margin

    return margins
```

```python
# Calculate gross margins
gross_margins = calculate_gross_margin(stock)

print("PRODUCT GROSS MARGINS")
print("-" * 40)

for product, margin in gross_margins.items():
    print(f"{product:<12} {margin:.2f}%")
```

```python
print("=" * 65)
print("             STOCK MANAGEMENT REPORT")
print("=" * 65)

print(f"{'PRODUCT':<12} {'QTY (KG)':>10} {'REORDER':>10} {'MARGIN':>12} {'STATUS':>12}")

print("-" * 65)

for product, details in stock.items():

    qty = details["qty"]
    reorder = details["reorder"]
    margin = gross_margins[product]

    if qty <= reorder:
        status = "Restock"
    else:
        status = "Sufficient"

    print(
        f"{product:<12} "
        f"{qty:>10} "
        f"{reorder:>10} "
        f"{margin:>11.2f}% "
        f"{status:>12}"
    )

print("=" * 65)
```
